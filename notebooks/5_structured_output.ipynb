### Structure Output

```python
from pydantic import BaseModel, Field
from typing import List

class ResponseStructure(BaseModel):
    name: str = Field(description='Full Name')
    email: str = Field(description='Email Address')
    age: int = Field(description='Age')
```

```python
structured_llm = llm.with_structured_output(ResponseStructure)
```

```python
resStruc = structured_llm.invoke(f'Please give me only name, email and age from this text: {text}')
```

```python
resStruc.name
```

```python
resStruc.email
```

```python
resStruc.age
```

```python
resStruc.model_dump()
```

```python
class Books(BaseModel):
    title: str = Field(description='Book Title')
    author: str = Field(description='Book Author')

class AllBooks(BaseModel):
    books: List[Books]
```

```python
book_llm = llm.with_structured_output(AllBooks)
```

```python
resAllBook = book_llm.invoke('Give me 3 books.')
```
