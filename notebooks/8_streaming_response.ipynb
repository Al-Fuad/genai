```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model='gpt-5-nano-2025-08-07', streaming=True)
```

```python
question = 'who is muhammad'
```

```python
res = llm.stream(question)
for chunk in res:
    print(chunk.content, end='')
```

```python
from langchain_groq import ChatGroq

llmGroq = ChatGroq(model='openai/gpt-oss-20b')
```

```python
resGroq = llmGroq.stream(question)
for chunk in resGroq:
    print(chunk.content, end='')
```
