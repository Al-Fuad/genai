```python
from langchain_unstructured import UnstructuredLoader
```

```python
text_path = '../data/speech.txt'
pdf_path = '../data/alfuad_resume.pdf'
```

```python
text_loader = UnstructuredLoader(file_path=text_path)
pdf_loader = UnstructuredLoader(file_path=pdf_path)
```

```python
texts = text_loader.load()
pdfs = pdf_loader.load()
```

```python
for text in texts:
    print(text.page_content)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
```

```python
text_chunks = splitter.split_documents(texts)
len(text_chunks)
```

```python
pdf_chunk = splitter.split_documents(pdfs)
len(pdf_chunk)
```
