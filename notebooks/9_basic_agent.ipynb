```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.tools import tool
```

```python
@tool
def add(a:int, b:int):
    """
        sum of two numbers

        Args:
            a: Number One
            b: Number Two
    """
    return a + b

@tool
def multiply(a:int, b:int):
    """
        product of two numbers

        Args:
            a: Number One
            b: Number Two
    """
    return a + b
```

```python
add.invoke({
    'a': 12,
    'b': 22
})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm = ChatOpenAI(model='gpt-5-nano-2025-08-07')
```

```python
agent = create_agent(
    model=llm,
    tools=[add, multiply],
    system_prompt='You are a math teacher, and always use tool from calculation'
)
```

```python
res = agent.invoke({
    'messages': [
        {
            'role': 'user',
            'content': 'add 1008 and 121 and multiply them with 9'
        }
    ]
})
```
