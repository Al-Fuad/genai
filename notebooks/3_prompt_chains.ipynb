#### Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
dynamicPrompts = ChatPromptTemplate.from_messages([
    {
        'role': 'system',
        'content': 'You are a translator and expert in {lang1} to {lang2} and {lang2} to {lang1}. translate the input to other languege'
    },
    {
        'role': 'user',
        'content': '{query}'
    }
])
dynamicPrompts
```

##### Function to print the prompt template

```python
def print_result(result: str):
    print(result.capitalize())
```

```python
final_prompt = dynamicPrompts.format_messages(
    lang1 = 'বাংলা',
    lang2 = 'Arabic',
    query = 'ختم الله على قلوبهم'
)
final_prompt
```

```python
res_dp = llm2.invoke(final_prompt)
```

```python
print(res_dp.content)
```

```python
res_dp = llm2.invoke(dynamicPrompts.invoke({
    'lang1': 'বাংলা',
    'lang2': 'arabic',
    'query': 'ختم الله على قلوبهم'
}))
```

```python
print(res_dp.content)
```

```python
chains = dynamicPrompts | llm2 | out | print_result
```

```python
resChains = chains.invoke({
    'lang1': 'English',
    'lang2': 'Banglish',
    'query': 'Who the hell are you?'
})
```
