```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents_to_embed: list[str] = [
    "Artificial intelligence is transforming modern software engineering.",
    "The quick brown fox jumps over the lazy dog.",
    "Vector databases allow for high-speed semantic search indexing.",
    "Data science involves statistical analysis and machine learning.",
    "Python is the preferred language for training neural networks."
]
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embeddings = OpenAIEmbeddings(model='text-embedding-3-small')
```

```python
vector = embeddings.embed_documents(documents_to_embed)
```

```python
len(vector[0])
```

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents_to_embed,
    embedding=embeddings,
    persist_directory='./vector_db'
)
```

```python
query = 'modern'
result = vector_store.similarity_search(query=query, k=2)
```
